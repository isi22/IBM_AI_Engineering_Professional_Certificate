## Introduction

The ability to extract precise answers from unstructured documents is a critical need across many industries, from legal and healthcare to finance and research. Traditional keyword-based search often falls short, struggling with context, nuance, and the sheer volume of information. This project addresses this challenge by developing a **Retrieval-Augmented Generation (RAG) powered Question-Answering (QA) chatbot** designed specifically to interact with PDF documents.

**Aim:**
The primary aim of this project is to create an intelligent chatbot that can answer user queries by retrieving relevant information directly from an uploaded PDF document and then generating coherent and contextually accurate responses using a Large Language Model (LLM). This significantly enhances the utility of document repositories, transforming them from static archives into interactive knowledge bases.

**Motivation:**
The motivation behind this project stems from the desire to make large volumes of textual data more accessible and queryable. Imagine a researcher needing to quickly find specific details across dozens of academic papers, or a legal professional cross-referencing clauses in lengthy contracts. Manually sifting through these documents is time-consuming and inefficient. An automated RAG-based system provides a powerful solution by:
* **Improving Information Retrieval:** Moving beyond simple keyword matches to understand the semantic meaning of queries.
* **Reducing Hallucinations:** Grounding LLM responses in factual information extracted from the provided document, thereby minimizing the generation of incorrect or fabricated details.
* **Enhancing User Experience:** Offering an intuitive conversational interface to interact with complex documents.

This project leverages the **LangChain** framework, a powerful toolkit for developing applications powered by language models, and integrates with **IBM's watsonx.ai platform** for its robust Large Language Model capabilities.

## Document Processing and Vectorization
This section handles transforming the raw PDF document into a format suitable for retrieval-augmented generation. It involves three main steps:
1.  **Loading the PDF:** Extracting text content from the uploaded PDF.
2.  **Text Chunking:** Dividing the extracted text into smaller, overlapping segments to manage context size for the LLM and improve retrieval accuracy.
3.  **Embedding Generation and Storage:** Converting these text chunks into numerical vector embeddings and storing them in a vector database for efficient similarity search.

### Define the PDF document loader

The `document_loader` function takes a PDF file path, loads it using `PyPDFLoader`, and returns a list of document objects. This function is the first step in preparing the document for the QA system.

```python
def document_loader(file):
    loader = PyPDFLoader(file.name)
    loaded_document = loader.load()
    return loaded_document
```

### Define the text splitter

The `text_splitter` function splits the loaded documents into smaller, manageable chunks. This is essential because LLMs have token limits, and smaller chunks allow for more precise retrieval and better context management. Overlapping chunks help ensure that context is not lost at the boundaries.

```python
def text_splitter(data, chunk_size, chunk_overlap):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len,
    )
    chunks = text_splitter.split_documents(data)
    return chunks
```

### Define the embedding model

The `WatsonxEmbedding` function configures the IBM Watsonx embedding model, which creates numerical vector embeddings for each text chunk. Embeddings capture the semantic meaning of text, allowing for efficient similarity searches based on the user's query.

```python
def WatsonxEmbedding(model_id, parameters):

    apikey_filename="watsonx_apikey.json"
    data = json.load(open(apikey_filename, 'r'))
    
    watsonx_embedding = WatsonxEmbeddings(
        model_id=model_id,
        url="https://eu-gb.ml.cloud.ibm.com",
        apikey=data.get("apikey"),
        project_id=data.get("project_id"), 
        params=parameters
    )

    return watsonx_embedding
```

### Define the vector store
The `vector_database` function creates a `Chroma` vector database. The vector store acts as the retrieval component in our RAG system.

```python
def vector_database(chunks, embedding_model):
    vectordb = Chroma.from_documents(chunks, embedding_model) # Create vector store from chunks and embeddings
    return vectordb
```

## Large Language Model (LLM) and Retrieval Chain Initialization
This section focuses on setting up the Large Language Model (LLM) that will be responsible for generating answers, and then combining it with the vector retrieval system into a cohesive Retrieval-Augmented Generation (RAG) chain.

### Initialize the LLM
The `watsonx_model` function configures the IBM Watsonx LLM. It's crucial to set up the API key, project ID, and model parameters like the model ID (e.g., `ibm/granite-13b-chat-v2`), temperature (creativity), and token limits (max_new_tokens) to ensure the LLM behaves as expected for the QA task.

```python
def watsonx_model(model_id, parameters=None):
    
    apikey_filename="watsonx_apikey.json"
    data = json.load(open(apikey_filename, 'r'))
    
    # To enable the LLM from watsonx.ai to work with LangChain, it needs to be converted to a chat model 
    mixtral_llm = WatsonxLLM(model_id=model_id,
                             params=parameters,
                             url="https://eu-gb.ml.cloud.ibm.com", # IBM Watsonx API endpoint
                             apikey=data.get("apikey"),
                             project_id=data.get("project_id")) 
    
    return mixtral_llm
```

### Setup RetrievalQA Chain
The `setup_qa_chain` function integrates the initialized LLM with the vector database to create a `RetrievalQA` chain. This chain takes a user query, retrieves relevant documents from the vector store, and then passes both the query and the retrieved context to the LLM to generate a final answer.

```python
def setup_qa_chain(llm, vectordb):
    qa_chain = RetrievalQA.from_chain_type(llm=llm, 
                                           chain_type="stuff", # "stuff" chain type puts all retrieved docs into prompt
                                           retriever=vectordb.as_retriever(), # Convert vector store to a retriever
                                           return_source_documents=False)
    return qa_chain
```

## Chatbot Interface with Gradio

### Define the retrieval chain

The `retriever_qa` function orchestrates the document loading, processing, and the QA chain execution. It handles file uploads and user queries to provide answers. 

```python
def retriever_qa(file, query, progress=gr.Progress()):

    # 1. Load document
    document_pdf = document_loader(file)
    print(f"Loaded {len(document_pdf)} pages from {file}") # Debugging print

    # 2. Chunk documents
        # Chunks of 1000 with an overlap of 50
    chunks = text_splitter(document_pdf, chunk_size=1000, chunk_overlap=50) 
    print(f"Split into {len(chunks)} chunks.") # Debugging print

    # 3. Create embeddings
        # IBM Slate 125M embedding model
    embed_params = {
        EmbedTextParamsMetaNames.TRUNCATE_INPUT_TOKENS: 3,
        EmbedTextParamsMetaNames.RETURN_OPTIONS: {"input_text": True},
    }
    embed_model_id="ibm/slate-125m-english-rtrvr" # Specify the embedding model to use
    watsonx_embedding = WatsonxEmbedding(embed_model_id, embed_params) # Initialize embedding model
    
    # 4. Create vector store
    vectordb = vector_database(chunks, watsonx_embedding)
    print("Embeddings created and stored in vectorstore.") # Debugging print

    # 5. Initialize LLM
        # Mixtral 8x7B LLM with a temperature of 0.5,and allowing for the generation of a maximum of 256 tokens
    llm_params = {
        GenParams.MAX_NEW_TOKENS: 256, # Maximum number of tokens in the generated response
        GenParams.TEMPERATURE: 0.5, # Controls creativity (lower for more factual, higher for more creative)
    }
    llm_model_id = 'mistralai/mixtral-8x7b-instruct-v01' # Specify the LLM model to use
    llm = watsonx_model(llm_model_id, llm_params) # Initialize LLM model
    print("LLM initialized.") # Debugging print

    # 6. Setup QA chain
    qa_chain = setup_qa_chain(llm, vectordb)
    print("QA chain setup complete.") # Debugging print

    # 7. Get answer from the QA chain
    response = qa_chain.invoke(query)
    print("Answer generated.") # Debugging print

    return response['result']
```

### Launch Gradio interface

This code block sets up and launches the Gradio web interface for the RAG chatbot. Users can upload a PDF, type their questions, and receive answers directly in their browser.
The `server_name` and `server_port` are configured for deployment flexibility, and `share=True` generates a public link for easy sharing (temporary).

```python
# Create the Gradio interface
rag_application = gr.Interface(
    fn=retriever_qa, # The function to call when the user interacts with the UI
    allow_flagging="never", # Disables Gradio's default flagging feature
    inputs=[
        gr.File(label="Upload PDF File", file_count="single", file_types=['.pdf'], type="filepath"),  # Drag and drop file upload
        gr.Textbox(label="Input Query", lines=2, placeholder="Type your question here...") # Input for user's text query
    ],
    outputs=gr.Textbox(label="Output"), # Output for the chatbot's answer
    title="RAG Chatbot", # Title displayed on the Gradio interface
    description="Upload a PDF document and ask any question. The chatbot will try to answer using the provided document." # Description below the title
)

# Launch the Gradio application
# server_name="0.0.0.0" allows access from external IPs (useful for deployment)
# server_port=7870 specifies the port to run on
# share=True generates a public URL (temporary, for easy sharing during development)
rag_application.launch(server_name="0.0.0.0", server_port= 7870, share=True)
```
